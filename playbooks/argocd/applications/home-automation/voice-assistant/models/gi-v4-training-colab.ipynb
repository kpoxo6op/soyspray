```python
# Run one durable boundary at a time: generate, augment, train, or finish.
TARGET = "generate"
```

```python
import pathlib
import shutil
import subprocess

SOYSPRAY_REVISION = "7854febf716a9927274e8231986890c19eeb2a29"
ROOT = pathlib.Path("/content/soyspray-gi-v4")

if TARGET not in {"generate", "augment", "train", "finish"}:
    raise RuntimeError(f"Unknown GI v4 target: {TARGET}")
if ROOT.exists():
    shutil.rmtree(ROOT)
ROOT.mkdir(parents=True)
subprocess.run(["git", "init", str(ROOT)], check=True)
subprocess.run(
    [
        "git",
        "-C", str(ROOT),
        "remote", "add", "origin",
        "https://github.com/kpoxo6op/soyspray.git",
    ],
    check=True,
)
subprocess.run(
    ["git", "-C", str(ROOT), "fetch", "--depth", "1", "origin", SOYSPRAY_REVISION],
    check=True,
)
subprocess.run(
    ["git", "-C", str(ROOT), "checkout", "--detach", SOYSPRAY_REVISION],
    check=True,
)
actual = subprocess.check_output(
    ["git", "-C", str(ROOT), "rev-parse", "HEAD"],
    text=True,
).strip()
if actual != SOYSPRAY_REVISION:
    raise RuntimeError(f"Wrong soyspray revision: {actual}")

subprocess.run(
    [
        "python3",
        str(ROOT / "scripts/gi_v4_browser_checkpoints.py"),
        "run", "--target", TARGET,
    ],
    check=True,
)
```
