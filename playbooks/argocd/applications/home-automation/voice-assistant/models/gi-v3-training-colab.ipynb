```python
import pathlib
import shutil
import subprocess

SOYSPRAY_REVISION = "c149a23c5297ce240486cf91b3a3adeefa6b38bd"
ROOT = pathlib.Path("/content/soyspray-gi-v3")

if ROOT.exists():
    shutil.rmtree(ROOT)
subprocess.run(
    [
        "git",
        "clone",
        "--branch", "codex/ha-voice-lights",
        "--single-branch",
        "--no-checkout",
        "https://github.com/kpoxo6op/soyspray.git",
        str(ROOT),
    ],
    check=True,
)
subprocess.run(
    ["git", "-C", str(ROOT), "checkout", "--detach", SOYSPRAY_REVISION],
    check=True,
)
actual = subprocess.check_output(
    ["git", "-C", str(ROOT), "rev-parse", "HEAD"],
    text=True,
).strip()
if actual != SOYSPRAY_REVISION:
    raise RuntimeError(f"Wrong soyspray revision: {actual}")

subprocess.run(
    [
        "python3",
        str(ROOT / "scripts/train_gi_v3_colab.py"),
        "--checkpoint-dir",
        "/content/drive/MyDrive/soyspray/home-assistant-voice/gi-v3",
    ],
    check=True,
)
```
